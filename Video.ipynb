```python
import cv2
```

```python
cap = cv2.VideoCapture(0)
if not cap.isOpened():
    print("Error")
    exit()
res = []
while True:
    ret, frame = cap.read()
    if not ret:
        print("Unable to grab frame")
        break
    cv2.imshow("Laptop Webcam", frame[:, ::-1, :])
    res.append(frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```
